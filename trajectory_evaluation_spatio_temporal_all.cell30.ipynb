```python
def get_reference_trajectory_input_tree(pv):
    ref_tree = {}
    
    for phone_os, phone_map in pv.map().items():
        for phone_label, phone_detail_map in phone_map.items():
            for (r_idx, r) in enumerate(phone_detail_map["evaluation_ranges"]):
                if r["eval_role_base"] != "accuracy_control":
                    continue
                for (tr_idx, tr) in enumerate(r["evaluation_trip_ranges"]):
                    for (sr_idx, sr) in enumerate(tr["evaluation_section_ranges"]):
                        # This is a Shapely LineString
                        # print(f"Retrieving ground truth for range {arrow.get(sr['start_ts'])} -> {arrow.get(sr['end_ts'])}")
                        section_gt_leg = pv.spec_details.get_ground_truth_for_leg(tr["trip_id_base"], sr["trip_id_base"], sr["start_ts"], sr["end_ts"])
                        section_gt_shapes = gpd.GeoSeries(eisd.SpecDetails.get_shapes_for_leg(section_gt_leg))
                        if len(section_gt_shapes) == 1:
                            print("No ground truth route for %s %s, must be polygon, skipping..." % (tr["trip_id_base"], sr["trip_id_base"]))
                            assert section_gt_leg["type"] != "TRAVEL", "For %s, %s, %s, %s, %s found type %s" % (phone_os, phone_label, r_idx, tr_idx, sr_idx, section_gt_leg["type"])
                            continue
                        if len(sr['location_df']) == 0:
                            print("No sensed locations found, role = %s skipping..." % (r["eval_role_base"]))
                            # assert r["eval_role_base"] == "power_control", "Found no locations for %s, %s, %s, %s, %s" % (phone_os, phone_label, r_idx, tr_idx, sr_idx)
                            continue
                            
                        print("Processing travel leg %s, %s, %s, %s, %s" %
                              (phone_os, phone_label, r["eval_role_base"], tr["trip_id_base"], sr["trip_id_base"]))
                        # print(f"Ground truth for trip range {arrow.get(tr['start_ts'])} and {arrow.get(tr['end_ts'])} is "
                        #       + f"START: {section_gt_leg['start_loc']['properties']['valid_start_fmt_date']} -> {section_gt_leg['start_loc']['properties']['valid_end_fmt_date']} "
                        #       + f"END:  {section_gt_leg['end_loc']['properties']['valid_start_fmt_date']} -> {section_gt_leg['end_loc']['properties']['valid_end_fmt_date']} ")
                        sec_name = tr["trip_id_base"] + "/" + sr["trip_id_base"] + "_" + str(r_idx)
                        if sec_name not in ref_tree:
                            ref_tree[sec_name] = {
                                "trip_id": tr["trip_id_base"],
                                "section_id": sr["trip_id_base"],
                                "run": r_idx,
                                "ground_truth": {
                                    "leg": section_gt_leg
                                }
                            }
                        
                        assert sec_name in ref_tree
                        e = ref_tree[sec_name]
                        # This is a GeoDataFrame
                        # section_measured_points = get_travel_trajectory(sr['location_df'], polygons)
                        section_measured_points = sr["location_df"]
                        if "temporal_control" not in e:
                            e["temporal_control"] = {}
                            e["start_ts"] = sr["start_ts"]
                            e["end_ts"] = sr["end_ts"]
                        e["temporal_control"][phone_os] = sr
    return ref_tree

def fill_ref_tree_entry(pv, e, function, include_ends=False):
    print("Considering entry %s %s %s and include_ends=%s" % (e["trip_id"], e["section_id"], e["run"], include_ends))
    curr_tz = pv.spec_details.eval_tz
    assert "android" in e["temporal_control"] and "ios" in e["temporal_control"]
    out = emr.ref_and_stats(e, function, 25, tz=curr_tz, include_ends=include_ends)
    if out is not None:
        e["reference_df"], e["reference_stats"] = out

def get_reference_trajectory_tree(pv, ref_tree_root, function, include_ends=False):
    curr_ref_tree = get_reference_trajectory_input_tree(pv)
    ref_tree_root[pv.spec_details.CURR_SPEC_ID] = curr_ref_tree
    [fill_ref_tree_entry(pv, e, function, include_ends) for e in curr_ref_tree.values()]
```

```python
def save_ref_tree(ref_tree, function, include_ends=False):
    for timeline in ref_tree.keys():
        print("Working with timeline %s" % timeline)
        tl = ref_tree[timeline]
        prefix = "with_ends" if include_ends else "no_ends"
        for trip_and_leg_and_run in tl.keys():
            curr_run = tl[trip_and_leg_and_run]
            fn = "bin/ref_trees/" + function + "/" + timeline+"/"+prefix+"/"+trip_and_leg_and_run
            save_dir = os.path.dirname(fn)
            os.makedirs(save_dir, exist_ok=True)
            print("Saving data to file %s" % fn)
            if "reference_df" not in curr_run:
                print("Found missing 'reference_df', skipping")
            else:
                curr_run.get('reference_df', pd.DataFrame()).to_csv(fn)
```

```python
import copy
import pickle
import geopandas as gpd
import shapely as shp
from joblib import Parallel, delayed
import arrow

print("FIRST TIMESTAMP: ",arrow.now())

# Python 3.11 still has the GIL, so the previous ThreadPoolExecutor gained
# nothing for this CPU-bound work. Switch to joblib's process-based parallelism
# (loky backend) so the builders actually run concurrently across cores.
N_JOBS = 100

# Each entry is (function, device). device=None means the existing two-stream
# references that merge android + ios. The single-stream and raw references are
# generated per device.
VARIANTS = [
    ("tf", None), ("ct", None), ("dtw", None),
    ("tf_single", "android"), ("tf_single", "ios"),
    ("dtw_single", "android"), ("dtw_single", "ios"),
    ("raw", "android"), ("raw", "ios"),
]

def variant_key(function, device):
    return function if device is None else "%s_%s" % (function, device)

def ref_fn(variant, timeline, sec_name, include_ends=True):
    prefix = "with_ends" if include_ends else "no_ends"
    return "bin/ref_trees/" + variant + "/" + timeline + "/" + prefix + "/" + sec_name

def stats_fn(variant, timeline, sec_name, include_ends=True):
    prefix = "with_ends" if include_ends else "no_ends"
    return "bin/ref_stats/" + variant + "/" + timeline + "/" + prefix + "/" + sec_name

pvs = [pv_la, pv_sj, pv_ucb]

# Build the (device-independent) input trees once per timeline. This is the
# expensive ground-truth/location loading; we do it serially and then fan out
# the per-variant reference construction across the process pool.
input_trees = {}      # timeline_id -> {sec_name: base_entry}
timeline_tz = {}      # timeline_id -> tz
for pv in pvs:
    input_trees[pv.spec_details.CURR_SPEC_ID] = get_reference_trajectory_input_tree(pv)
    timeline_tz[pv.spec_details.CURR_SPEC_ID] = pv.spec_details.eval_tz

# Pre-serialize every section's base_entry exactly once. base_entry holds the
# raw android/ios location dataframes and is expensive to pickle, so doing it
# here means the same bytes are reused for every variant task that operates on
# that section instead of joblib re-pickling the live object once per task.
serialized_entries = {}   # timeline_id -> {sec_name: pickled base_entry bytes}
for timeline, tree_in in input_trees.items():
    serialized_entries[timeline] = {
        sec_name: pickle.dumps(base_entry, protocol=pickle.HIGHEST_PROTOCOL)
        for sec_name, base_entry in tree_in.items()
    }

def save_one_reference(variant, timeline, sec_name, reference_df, reference_stats, include_ends=True):
    # Persist a single reference trajectory + its stats, mirroring the directory
    # layout used by save_ref_tree:
    #   bin/ref_trees/<variant>/<timeline>/<with_ends|no_ends>/<sec_name>
    #   bin/ref_stats/<variant>/<timeline>/<with_ends|no_ends>/<sec_name>
    fn = ref_fn(variant, timeline, sec_name, include_ends)
    sfn = stats_fn(variant, timeline, sec_name, include_ends)
    os.makedirs(os.path.dirname(fn), exist_ok=True)
    os.makedirs(os.path.dirname(sfn), exist_ok=True)
    reference_df.to_csv(fn)
    pd.DataFrame(reference_stats, index=[0]).to_csv(sfn)

def load_reference_df(fn):
    # Reconstruct a reference GeoDataFrame from its saved CSV. The geometry
    # column is rebuilt from the persisted latitude/longitude so downstream
    # code that relies on `reference_df.geometry` keeps working.
    df = pd.read_csv(fn)
    return gpd.GeoDataFrame(
        df,
        geometry=df.apply(lambda lr: shp.geometry.Point(lr.longitude, lr.latitude), axis=1))

def build_and_save_reference(function, device, timeline, sec_name, base_entry_bytes, tz, dtw_ref_df_bytes=None):
    # Runs in a worker process. It unpickles the shared section bytes into a
    # private entry (so workers never mutate shared state), builds the variant,
    # saves it straight to disk, and returns nothing -- the parent rebuilds
    # ref_trees from the saved files afterwards.
    e = pickle.loads(base_entry_bytes)
    out = emr.ref_and_stats(e, function, 25, tz=tz, include_ends=True, device=device)
    if out is None:
        return
    reference_df, reference_stats = out
    # For the single-stream / raw (device-specific) variants, fold the
    # spatio-temporal error against the already-built two-stream DTW reference
    # for the same segment into reference_stats. The DTW reference df is built
    # in phase 1 and passed in as bytes, so we never rebuild it here.
    if device is not None and dtw_ref_df_bytes is not None:
        dtw_ref_df = pickle.loads(dtw_ref_df_bytes)
        reference_stats["st_error_vs_dtw"] = emr.spatio_temporal_error(dtw_ref_df, reference_df)
    save_one_reference(variant_key(function, device), timeline, sec_name, reference_df, reference_stats)

# Phase 1: build the two-stream DTW references first. The single-stream / raw
# variants depend on them to compute the spatio-temporal error, so this batch
# must finish (and be reloadable from disk) before phase 2 starts.
dtw_tasks = []
for timeline, tree_in in input_trees.items():
    for sec_name in tree_in:
        dtw_tasks.append(("dtw", None, timeline, sec_name,
                          serialized_entries[timeline][sec_name], timeline_tz[timeline], None))

print("Phase 1: building %d two-stream DTW references on %d processes" % (len(dtw_tasks), N_JOBS))
Parallel(n_jobs=N_JOBS, verbose=10)(delayed(build_and_save_reference)(*t) for t in dtw_tasks)

# Reload each DTW reference from disk and pre-serialize it so phase-2 device
# tasks can reuse the same bytes for the spatio-temporal error calc.
dtw_ref_bytes = {}    # (timeline, sec_name) -> pickled dtw reference_df bytes
for timeline, tree_in in input_trees.items():
    for sec_name in tree_in:
        fn = ref_fn("dtw", timeline, sec_name)
        if os.path.exists(fn):
            dtw_ref_bytes[(timeline, sec_name)] = pickle.dumps(
                load_reference_df(fn), protocol=pickle.HIGHEST_PROTOCOL)

# Phase 2: everything else (tf, ct, and the single-stream / raw variants).
other_tasks = []
for function, device in VARIANTS:
    if (function, device) == ("dtw", None):
        continue
    for timeline, tree_in in input_trees.items():
        for sec_name in tree_in:
            d_bytes = dtw_ref_bytes.get((timeline, sec_name)) if device is not None else None
            other_tasks.append((function, device, timeline, sec_name,
                                serialized_entries[timeline][sec_name], timeline_tz[timeline], d_bytes))

print("SECOND TIMESTAMP: ",arrow.now())
print("Phase 2: building %d remaining references on %d processes" % (len(other_tasks), N_JOBS))
Parallel(n_jobs=N_JOBS, verbose=10)(delayed(build_and_save_reference)(*t) for t in other_tasks)
print("THIRD TIMESTAMP: ",arrow.now())
# Reconstruct ref_trees from the files the workers wrote. Each entry starts from
# the in-memory base_entry (so ground_truth / temporal_control are available for
# the downstream viz cells) and gets reference_df / reference_stats loaded back
# from disk. fill_gt_linestring restores the ground-truth linestring that
# ref_and_stats used to add as a side effect in the old single-process flow.
# ref_trees[variant_key][timeline_id][sec_name] = entry
ref_trees = {variant_key(f, d): {tl: {} for tl in input_trees} for f, d in VARIANTS}
for function, device in VARIANTS:
    vkey = variant_key(function, device)
    for timeline, tree_in in input_trees.items():
        for sec_name, base_entry in tree_in.items():
            fn = ref_fn(vkey, timeline, sec_name)
            sfn = stats_fn(vkey, timeline, sec_name)
            if not os.path.exists(fn):
                continue
            e = copy.copy(base_entry)
            e["ground_truth"] = {"leg": base_entry["ground_truth"]["leg"]}
            emr.fill_gt_linestring(e)
            e["reference_df"] = load_reference_df(fn)
            e["reference_stats"] = pd.read_csv(sfn).iloc[0].to_dict()
            ref_trees[vkey][timeline][sec_name] = e

print("Built variants: %s" % list(ref_trees.keys()))

# Backward-compatible list for the existing two-stream analysis cells below.
# Order matches the downstream assumption: [0]=tf, [1]=ct, [2]=dtw.
ref_trees_all = [ref_trees["tf"], ref_trees["ct"], ref_trees["dtw"]]

# Backward-compatible aliases used by the downstream visualization cells, which
# expect a {timeline_id: {sec_name: entry}} tree. Use the dtw tree (matches the
# previous behaviour where ref_tree_with_ends ended as the last-built tree).
ref_tree_with_ends = ref_trees["dtw"]
ref_tree = ref_tree_with_ends

with open("ref_trees_all.pkl", 'wb') as f:
    pickle.dump(ref_trees_all, f)
# Save the full variant dict (includes single-stream and raw references) too.
with open("ref_trees.pkl", 'wb') as f:
    pickle.dump(ref_trees, f)

print("FOURTH TIMESTAMP: ",arrow.now())
```
